```python
import pandas as pd
import matplotlib.pyplot as plt
import csv
import mysql.connector
conn=mysql.connector.connect(host="localhost",user="root",password="password")
cursor=conn.cursor()
```

```python
cursor.execute("create database if not exists smartcity")
cursor.execute("use smartcity")
```

```python
with open (r"C:\Users\akshi\OneDrive\Documents\jupyter notebook\Smart_City_All_Modules_Merged_Complete.csv",mode="r",encoding="utf-8-sig") as f:
    reader=csv.reader(f)
    headers=next(reader)
    cleaned_headers=[h.strip().replace(" ","_") for h in headers]
Columns_def=", ".join(f"`{col}` TEXT" for col in cleaned_headers)
cursor.execute(f"CREATE TABLE IF NOT EXISTS smart_city ({Columns_def})")
placeholders=", ".join(["%s"] * len(cleaned_headers))
insert_query = f"INSERT INTO smart_city VALUES ({placeholders})"
with open (r"C:\Users\akshi\OneDrive\Documents\jupyter notebook\Smart_City_All_Modules_Merged_Complete.csv",mode="r",encoding="utf-8-sig") as f:
    reader=csv.reader(f)
    next(reader)
    rows_to_insert = [tuple(val if val.strip() != "" else None for val in row) for row in reader]
cursor.executemany(insert_query, rows_to_insert)
conn.commit()

cursor.close()
conn.close()

print(" Success! Your traffic data has been imported into the 'smartcity' database.")
```
